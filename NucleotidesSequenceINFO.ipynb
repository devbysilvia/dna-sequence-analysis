```python
def count_nucleotides(sequence):
    counts = {
        "A": sequence.count("A"),
        "T": sequence.count("T"),
        "C": sequence.count("C"),
        "G": sequence.count("G")
    }
    total_length = len(sequence)
    if total_length == 0:
        return 0
    count_A = sequence.count("A")
    count_T = sequence.count("T")
    count_C = sequence.count("C")
    count_G = sequence.count("G")
    gc_content = ((count_C + count_G)/total_length)*100
    if gc_content < 40:
        gc_classification = "Low"
    elif 40 <= gc_content <= 60:
        gc_classification = "Medium"
    else:
        gc_classification = "High"
    return counts, total_length, gc_content, gc_classification
```

```python
sequence1 = "ATGCGTAGCTAG"
counts, total, gc, gc_classification = count_nucleotides(sequence1)

print(f"Sequence length: {total} bases")
print(f"GC content: {gc:.2f}% ({gc_classification})")
for base, count in counts.items():
    print(f"Count of {base}: {count} nucleotides")
```
