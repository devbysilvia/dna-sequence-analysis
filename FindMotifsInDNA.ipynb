```python
def find_motif(sequence, motif):
    positions = [] # we create a list that will include the found motives
    start = 0 # we set start at position 0
    while start < len(sequence):
        pos = sequence.find(motif, start) #we start searching motives at start (0)
        if pos != -1: # if there are no found motives, it will return '-1'. If found, add motif to list positions
            positions.append(pos)
            start = pos + 1 # if found motif, search at next position
        else:
            break 
    
        
    return positions
```

```python
sequence = "ATGCGTAGCTAGATGCTAGATG"
motif = "ATG"
positions = find_motif(sequence, motif)
print(f"The motif 'ATG' is found at positions: {positions}")
```
